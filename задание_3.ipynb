```python
region_code = "76"
salary_level = 30000
# Список для сохранения всех найденных вакансий
results_list = []

for offset_number in tqdm(range(0, 10)):
    link = f"http://opendata.trudvsem.ru/api/v1/vacancies/region/{region_code}?offset={offset_number}"
    anws = requests.get(link)
    time.sleep(1)

    if anws.status_code == 200:
        dct_region = anws.json()

        results = dct_region.get("results") or {}
        lst_vac = results.get("vacancies") or []

        for vac in lst_vac:
            vacancy_data = vac.get("vacancy") or {}

            raw_salary = vacancy_data.get("salary", 0)
            try:
                salary = int(raw_salary) if raw_salary is not None else 0
            except (ValueError, TypeError):
                salary = 0

            if salary >= salary_level:
                company = vacancy_data.get("company") or {}

                addresses = vacancy_data.get("addresses") or {}
                address_list = addresses.get("address") or []
                location = (
                    address_list[0].get("location", "Адрес не указан")
                    if address_list
                    else "Адрес не указан"
                )

                # Создаем словарь под каждую вакансию
                dct_one_vac = {
                    "Название региона": vacancy_data.get("region", {}).get(
                        "name", "Название региона не указано"
                    ),
                    "Название компании": company.get(
                        "name", "Компания не указана"
                    ),
                    "Почта компании": company.get(
                        "email", "Почта не указана"
                    ),
                    "ИНН": company.get("inn", "ИНН не указан"),
                    "Вакансия": vacancy_data.get(
                        "job-name", "Название не указано"
                    ),
                    "Предлагаемая зарплата": salary,
                    "Обязанности": vacancy_data.get(
                        "duty", "Обязанности не указаны"
                    ),
                    "Адрес компании": location,
                }

                # Сохраняем словарь в общий список
                results_list.append(dct_one_vac)
```

```python
# Превращаем результат в DataFrame
df = pd.DataFrame(results_list)

df.head()
```

**Задача 2. Web-scraping**

```python
from bs4 import BeautifulSoup
```

```python
# формируем запрос
page_number = 2
req = f'https://www.piter.com/collection/kompyutery-i-internet?page={page_number}'
```

```python
# отправка запроса
anws = requests.get(req)

anws.status_code
```

```python
# анализ ответа
anws.text
```

```python
# разбор ответа
soup = BeautifulSoup(anws.text, 'lxml')
soup
```

```python
# исследование страницы
results = soup.find_all('div',
                        attrs={'class': ['grid-3 prod-block book-block clear-class2 clear-class4',
                                         'grid-3 prod-block book-block',
                                         'grid-3 prod-block book-block clear-class2']})
len(results)
```

```python
# сбор данных
dct_book = {'Автор': None,
            'Название': None,
            'Ссылка': None}
lst_books = []

for book in results:
    dct_book['Автор'] = (
                                book.find('span',
                                          attrs={'class': 'author'})
                                    .get_text(strip=True)
                                )
    dct_book['Название'] = (
                                book.find('span',
                                          attrs={'class': 'title'})
                                    .get_text(strip=True)
                                  )
    dct_book['Ссылка'] = (
                     'https://www.piter.com' + book.find('a').get('href')
                         )
    lst_books.append(dct_book.copy())
```

```python
pd.DataFrame(lst_books)
```
