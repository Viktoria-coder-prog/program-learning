```python
# Задача 11. Добавьте новый категориальный признак bodyType на основе столбца BMI:
# BMI Categories: Underweight = <18.5; Normal weight = 18.5–24.9; Overweight = 25–29.9; Obesity = BMI of 30 or greater.
# Признак должен принимать значения Underweight, Normal weight, Overweight и Obesity.
ctgrs = [
    (df['bmi'] < 18.5),
    (df['bmi'] >= 18.5) & (df['bmi'] <= 24.9),
    (df['bmi'] >= 25) & (df['bmi'] <= 29.9),
    (df['bmi'] >= 30)
]

column = [
    "Underweight",
    "Normal weight",
    "Overweight",
    "Obesity"
]

df['bmi_categories'] = np.select(ctgrs, column, default="Unknown")
df[['bmi', 'bmi_categories']]
```

```python
# Задача 12. Будем считать "здоровыми" тех, у кого нормальный вес и кровяное давление.
# Какой процент "здоровых" женщин больны диабетом?
normal_weight = df['bmi'].between(18.5, 24.9)
normal_pressure = df['bloodpressure'].between(80, 89)
healthy_fltr = normal_weight & normal_pressure
diabetes_percent = df[healthy_fltr]['class'].mean() * 100
print(diabetes_percent)
```
